# Best Test and General Skills Quiz

**Question:** Calculate the sum of the following quantiles and round to 2 decimal places:

- Chi-square quantile: 99th percentile with 9 degrees of freedom
- t-distribution quantile: 90th percentile with 6 degrees of freedom

**Formula:** $\chi^2_{0.01,9} + t_{0.90,6}$

**Instructions:** Use R functions `qchisq()` and `qt()` to find the quantiles, then use `round()` to round the result to 2 decimal places.

```r
#correct answer for Best Test and Some General Skills Quiz
round(qchisq(0.99, 9) + qt(0.90, 6), 2)
```
